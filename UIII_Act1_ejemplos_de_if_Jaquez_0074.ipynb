#### Ejemplos de if, if else, if elif

```python
#1. Sentencia if simple (Uso de operadores aritméticos y relacionales)
#Este tipo de estructura evalúa una condición única; si es verdadera, ejecuta el bloque de código indentado. Si es falsa, el programa simplemente continúa con la siguiente instrucción fuera del bloque
numero = 15 # Asigna el valor 15 a la variable [8, 9]
if numero % 3 == 0: # El operador aritmético '%' obtiene el residuo; '==' compara si es igual a 0 [10-12]
    print("El número es múltiplo de 3") # Imprime el mensaje si la condición es verdadera [8, 13]
2. Estructura if else (Uso de operadores lógicos y relacionales)
```

```python
#2. Estructura if else (Uso de operadores lógicos y relacionales)
#La estructura if-else permite definir dos caminos: uno que se sigue si la condición se cumple (True) y otro (el else) que se ejecuta únicamente si la condición inicial no se cumple (False)
.
edad = 20 # Define la variable edad con un valor entero [16, 17]
tiene_entrada = True # Define una variable booleana de estado [17, 18]
if edad >= 18 and tiene_entrada: # '>=' es un operador relacional; 'and' es un operador lógico que requiere que ambos sean True [11, 19-21]
    print("Acceso concedido") # Se ejecuta si ambas condiciones combinadas por el 'and' son verdaderas [7, 22]
else: # Bloque que actúa como alternativa cuando el 'if' resulta falso [13, 16, 23]
    print("Acceso denegado") # Mensaje de salida para el caso en que no se cumpla la condición [16, 24]
```

```python
#3. Estructura if elif else (Uso de múltiples condiciones y comparaciones)
#Cuando existen más de dos posibilidades, se utiliza elif (abreviatura de else if). Python evalúa las condiciones en orden y, en cuanto encuentra una verdadera, ejecuta su bloque y sale de toda la estructura condicional
nota = int(input("Ingresa tu nota del 0 al 100")) # Almacena la calificación a evaluar [25, 26]
if nota >= 90: # Verifica si la nota es mayor o igual a 90 utilizando el operador relacional '>=' [19, 25]
    print("Calificación: A") # Se imprime si la primera condición es True [25, 26]
elif nota >= 80: # Si la primera fue False, verifica si es mayor o igual a 80 [25, 28]
    print("Calificación: B") # Se imprime si solo esta condición es verdadera [26, 27]
elif nota >= 70: # Evalúa una tercera opción si las anteriores fallaron [25, 27]
    print("Calificación: C") # Muestra la calificación correspondiente a este rango [25]
else: # Se ejecuta si ninguna de las condiciones anteriores (if o elif) fue verdadera [25, 26, 28]
    print("No aprobado") # Caso por defecto para notas menores a 70 [25, 26]
```

```python
#4. Condicionales anidados (Lógica compleja combinada)
#Es posible incluir un if dentro de otro. Esto es útil para verificaciones que dependen del éxito de una condición previa
usuario_registrado = True # Indica si el usuario existe en el sistema [17]
contrasena_correcta = "Admin123" # Valor de referencia para la clave [32]
entrada_usuario = "Admin123" # Simula la clave ingresada por el usuario [6]
if usuario_registrado: # Comprueba la primera condición de existencia del usuario [6, 7]
    if entrada_usuario == contrasena_correcta: # Condicional anidado que usa '==' para comparar igualdad exacta [30, 33]
        print("Sesión iniciada") # Se ejecuta si el usuario existe Y la contraseña es correcta [30]
    else: # Alternativa si el usuario existe pero falló la contraseña [34, 35]
        print("Contraseña incorrecta") # Mensaje de error específico de seguridad [36]
else: # Alternativa si la primera condición (usuario_registrado) fue falsa [13, 16]
    print("El usuario no existe") # Mensaje informativo para usuarios no registrados [37]
```
